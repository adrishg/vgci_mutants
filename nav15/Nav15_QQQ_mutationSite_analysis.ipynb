# Nav1.5 IFM→QQQ regional distance analysis
QQQ replaces the I1485-F1486-M1487 fast-inactivation motif with Q1485-Q1486-Q1487. This notebook tests its distal effect on the selectivity-filter region (D373/E704/K1103/A1395); these residues are not the mutation site. Vanilla, masked, and masked-v2 QQQ ensembles are compared with experimental distances from 7FBS, 6UZ3, 8VYJ, and 8VYK.


```python
from pathlib import Path
import importlib
import sys
import pandas as pd
repo_root = Path.cwd() if (Path.cwd() / 'shared').is_dir() else Path.cwd().parent
sys.path.insert(0, str(repo_root)) if str(repo_root) not in sys.path else None
import shared.plotting as shared_plotting
importlib.reload(shared_plotting)  # Pick up new plotting helpers in an existing kernel.
plot_distances_by_alias_violin = shared_plotting.plot_distances_by_alias_violin
plot_protocol_split_with_experimentals = shared_plotting.plot_protocol_split_with_experimentals
import shared.dataset_selection as dataset_selection
importlib.reload(dataset_selection)
distance_csv_options = dataset_selection.distance_csv_options
load_selected_distance_csv = dataset_selection.load_selected_distance_csv
NAV15_PALETTE = {'QQQ_VAN':'#E6D0F2','QQQ_HM':'#9268BB'}
```

```python
QQQ_FILTER_ALIASES = {
    'D373-E704': 'CA_ASP373_CA-GLU704_CA',
    'D373-K1103': 'CA_ASP373_CA-LYS1103_CA',
    'D373-A1395': 'CA_ASP373_CA-ALA1395_CA',
    'E704-K1103': 'CA_GLU704_CA-LYS1103_CA',
    'E704-A1395': 'CA_GLU704_CA-ALA1395_CA',
    'K1103-A1395': 'CA_LYS1103_CA-ALA1395_CA',
}
QQQ_EXPERIMENTAL_DISTANCES = {
    '7FBS | QQQ mutant': {
        'D373-E704': [5.804], 'D373-K1103': [10.912], 'D373-A1395': [8.988],
        'E704-K1103': [10.342], 'E704-A1395': [12.075], 'K1103-A1395': [6.840],
    },
    '6UZ3 | WT': {
        'D373-E704': [5.332], 'D373-K1103': [9.773], 'D373-A1395': [8.780],
        'E704-K1103': [9.433], 'E704-A1395': [11.710], 'K1103-A1395': [6.371],
    },
    '8VYJ | Class I': {
        'D373-E704': [6.252], 'D373-K1103': [11.170], 'D373-A1395': [8.817],
        'E704-K1103': [10.645], 'E704-A1395': [12.573], 'K1103-A1395': [7.742],
    },
    '8VYK | Class II': {
        'D373-E704': [6.131], 'D373-K1103': [10.598], 'D373-A1395': [8.243],
        'E704-K1103': [10.589], 'E704-A1395': [12.419], 'K1103-A1395': [7.787],
    },
}
QQQ_EXPERIMENTAL_DISTANCES
```

```python
# Load every QQQ protocol so the two masking strategies are compared directly.
DATASET_SELECTION = 'all_ok_3'
data = repo_root / 'nav15' / 'dataDistances'
DATASET_PATHS = {
    'QQQ vanilla': distance_csv_options(repo_root, data / '26-02-01_Nav15_qqq_vanilla_distances_all.csv', 'Nav15', 'QQQ', 'vanilla'),
    'QQQ masked': distance_csv_options(repo_root, data / '26-07-24_Nav15_qqq_maskedAF2_distances_all.csv', 'Nav15', 'QQQ', 'masked'),
    'QQQ masked_v2': distance_csv_options(repo_root, data / '26-07-25_Nav15_qqq_maskedAF2v2_distances_all.csv', 'Nav15', 'QQQ', 'masked'),
}
DATASET_PATHS
df_vanilla = load_selected_distance_csv('QQQ vanilla', DATASET_PATHS['QQQ vanilla'], DATASET_SELECTION)
df_masks = {
    variant: load_selected_distance_csv(f'QQQ {variant}', DATASET_PATHS[f'QQQ {variant}'], DATASET_SELECTION)
    for variant in ('masked', 'masked_v2')
}
experimental_list = list(QQQ_EXPERIMENTAL_DISTANCES.values())
experimental_labels = [f'Experimental | {label}' for label in QQQ_EXPERIMENTAL_DISTANCES]
experimental_colors = ['#D55E00','#0072B2','#009E73','#CC79A7']
for protocol, frame, color in [('vanilla', df_vanilla, NAV15_PALETTE['QQQ_VAN']), ('masked', df_masks['masked'], '#B69AD0'), ('masked_v2', df_masks['masked_v2'], NAV15_PALETTE['QQQ_HM'])]:
    plot_distances_by_alias_violin(frame, QQQ_FILTER_ALIASES, exp_distances_list=experimental_list, title_custom_add=f'Nav1.5 | IFM→QQQ | {protocol} | selectivity-filter region | experimental distances', colors=[color] * len(QQQ_FILTER_ALIASES), fig_width=10.5, fig_height=8.0, dataset_labels=experimental_labels, pdb_colors=experimental_colors)

plot_protocol_split_with_experimentals(
    df_masks['masked'], df_masks['masked_v2'], QQQ_FILTER_ALIASES,
    title='Nav1.5 | IFM→QQQ | masked vs masked v2 | selectivity-filter region | split violin | experimental distances',
    colors=['#B69AD0', NAV15_PALETTE['QQQ_HM']],
    exp_distances_list=experimental_list, dataset_labels=experimental_labels, pdb_colors=experimental_colors,
)

# Compare vanilla against each masking strategy as separate, explicit tests.
for mask_name, mask_frame, mask_color in [
    ('masked', df_masks['masked'], '#B69AD0'),
    ('masked v2', df_masks['masked_v2'], NAV15_PALETTE['QQQ_HM']),
]:
    plot_protocol_split_with_experimentals(
        df_vanilla, mask_frame, QQQ_FILTER_ALIASES,
        title=f'Nav1.5 | IFM→QQQ | vanilla vs {mask_name} | selectivity-filter region | split violin | experimental distances',
        colors=[NAV15_PALETTE['QQQ_VAN'], mask_color],
        exp_distances_list=experimental_list, dataset_labels=experimental_labels, pdb_colors=experimental_colors,
    )
```
