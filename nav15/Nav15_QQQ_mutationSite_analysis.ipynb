# $\mathrm{Na}_{\mathrm{V}}1.5$ IFM→QQQ regional distance analysis
QQQ replaces the I1485-F1486-M1487 fast-inactivation motif with Q1485-Q1486-Q1487. This notebook tests its distal effect on the selectivity-filter region (D373/E704/K1103/A1395); these residues are not the mutation site. Vanilla, masked, and masked-v2 QQQ ensembles are compared with experimental distances from 7FBS, 6UZ3, 8VYJ, and 8VYK.


## Comparison design

The six Cα distances describe the four-residue selectivity-filter geometry, not
the IFM/QQQ mutation site itself. Vanilla is compared separately with the
original mask and masked v2, and the two masked protocols are also compared
directly. Experimental points from 7FBS, 6UZ3, 8VYJ, and 8VYK provide pore-state
context without being treated as QQQ-latch measurements.

## Selectivity-filter distributions

Each split violin retains the same residue-pair order so that protocol effects
can be compared directly. Differences in these distal pore coordinates indicate
an associated change in filter geometry, but they do not identify the IFM
engagement state; that question is addressed in the dedicated latching
notebook.

```python
# Load every QQQ protocol so the two masking strategies are compared directly.
DATASET_SELECTION = 'all_ok_3'
data = repo_root / 'nav15' / 'dataDistances'
DATASET_PATHS = {
    'QQQ vanilla': distance_csv_options(repo_root, data / '26-07-27_Nav15_qqq_vanilla_AF2_distances_all_all.csv', 'Nav15', 'QQQ', 'vanilla'),
    'QQQ masked': distance_csv_options(repo_root, data / '26-07-27_Nav15_qqq_masked_AF2_distances_all_all.csv', 'Nav15', 'QQQ', 'masked'),
    'QQQ masked_v2': distance_csv_options(repo_root, data / '26-07-27_Nav15_qqq_maskedv2_AF2_distances_all_all.csv', 'Nav15', 'QQQ', 'masked'),
}
DATASET_PATHS
df_vanilla = load_selected_distance_csv('QQQ vanilla', DATASET_PATHS['QQQ vanilla'], DATASET_SELECTION)
df_masks = {
    variant: load_selected_distance_csv(f'QQQ {variant}', DATASET_PATHS[f'QQQ {variant}'], DATASET_SELECTION)
    for variant in ('masked', 'masked_v2')
}
experimental_list = list(QQQ_EXPERIMENTAL_DISTANCES.values())
experimental_labels = [f'Experimental | {label}' for label in QQQ_EXPERIMENTAL_DISTANCES]
experimental_colors = ['#D55E00','#0072B2','#009E73','#CC79A7']
for protocol, frame, color in [('vanilla', df_vanilla, NAV15_PALETTE['QQQ_VAN']), ('masked', df_masks['masked'], NAV15_PALETTE['QQQ_MASKED']), ('masked_v2', df_masks['masked_v2'], NAV15_PALETTE['QQQ_MASKED_V2'])]:
    plot_distances_by_alias_violin(frame, QQQ_FILTER_ALIASES, exp_distances_list=experimental_list, title_custom_add=f'Nav1.5 | IFM→QQQ | {protocol} | selectivity-filter region | experimental distances', colors=[color] * len(QQQ_FILTER_ALIASES), fig_width=10.5, fig_height=8.0, dataset_labels=experimental_labels, pdb_colors=experimental_colors)

plot_protocol_split_with_experimentals(
    df_masks['masked'], df_masks['masked_v2'], QQQ_FILTER_ALIASES,
    title='Nav1.5 | IFM→QQQ | masked vs masked v2 | selectivity-filter region | split violin | experimental distances',
    colors=[NAV15_PALETTE['QQQ_MASKED'], NAV15_PALETTE['QQQ_MASKED_V2']],
    exp_distances_list=experimental_list, dataset_labels=experimental_labels, pdb_colors=experimental_colors,
)

# Compare vanilla against each masking strategy as separate, explicit tests.
for mask_name, mask_frame, mask_color in [
    ('masked', df_masks['masked'], NAV15_PALETTE['QQQ_MASKED']),
    ('masked v2', df_masks['masked_v2'], NAV15_PALETTE['QQQ_MASKED_V2']),
]:
    plot_protocol_split_with_experimentals(
        df_vanilla, mask_frame, QQQ_FILTER_ALIASES,
        title=f'Nav1.5 | IFM→QQQ | vanilla vs {mask_name} | selectivity-filter region | split violin | experimental distances',
        colors=[NAV15_PALETTE['QQQ_VAN'], mask_color],
        exp_distances_list=experimental_list, dataset_labels=experimental_labels, pdb_colors=experimental_colors,
    )

## Shortest-distance contacts of the three QQQ residues
from shared.mutation_contact_analysis import plot_mutation_contacts, plot_selected_contact_distributions
WT_CONTACT_PATHS = {
    'vanilla': distance_csv_options(repo_root, data / '26-07-27_Nav15_wt_vanillaAF2_distances_all_all.csv', 'Nav15', 'WT', 'vanilla'),
    'masked_v2': distance_csv_options(repo_root, data / '26-07-27_Nav15_wt_maskedv2_AF2_distances_all_all.csv', 'Nav15', 'WT', 'masked'),
}
wt_contact_frames = {
    protocol: load_selected_distance_csv(f'WT {protocol}', paths, DATASET_SELECTION)
    for protocol, paths in WT_CONTACT_PATHS.items()
}
qqq_contact_comparisons = {
    'vanilla': (wt_contact_frames['vanilla'], df_vanilla),
    'masked v2': (wt_contact_frames['masked_v2'], df_masks['masked_v2']),
}
qqq_filter_shift_tables = {
    protocol: distance_shift_table(wt, qqq, QQQ_FILTER_ALIASES)
    for protocol, (wt, qqq) in qqq_contact_comparisons.items()
}
for protocol, shifts in qqq_filter_shift_tables.items():
    plot_top_shifts(
        shifts,
        f'Largest Nav1.5 selectivity-filter distance shifts: IFM→QQQ {protocol}',
        n=len(QQQ_FILTER_ALIASES),
    )
QQQ_POSITION_MAP = [('ILE1169', 'GLN1169', 'I1169Q'), ('PHE1170', 'GLN1170', 'F1170Q'), ('MET1171', 'GLN1171', 'M1171Q')]
qqq_contact_tables = {}
for wt_residue, q_residue, position_label in QQQ_POSITION_MAP:
    table, _ = plot_mutation_contacts(
        qqq_contact_comparisons, wt_residue, q_residue, position_label,
        {'vanilla': NAV15_PALETTE['QQQ_VAN'], 'masked v2': NAV15_PALETTE['QQQ_MASKED_V2']}, channel='Nav1.5',
    )
    qqq_contact_tables[position_label] = table
    display(table.head(12).round(3))
    top_partners = list(dict.fromkeys(table['Partner']))[:3]
    plot_selected_contact_distributions(
        qqq_contact_comparisons, wt_residue, q_residue, position_label, top_partners,
        {'vanilla': (NAV15_PALETTE['WT_VAN'], NAV15_PALETTE['QQQ_VAN']), 'masked v2': (NAV15_PALETTE['WT_MASKED_V2'], NAV15_PALETTE['QQQ_MASKED_V2'])}, channel='Nav1.5',
    )
```
